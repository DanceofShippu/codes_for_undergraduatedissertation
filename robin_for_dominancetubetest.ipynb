```python
#(pseudorandomness)single tournament of four mice, taking sides(left and right) into consideration
#created by Gangning Liu
import random

mice_list=[["A","A","A"],["B","B","B"],["C","C","C"],["D","D","D"]]
counter=0
competition_list=[]
while counter!=6:
    index1=random.randint(0,3)
    while mice_list[index1]==[]:
        index1=random.randint(0,3)
    mouse1=mice_list[index1].pop()
    index2=random.randint(0,3)
    while (index2==index1) or (mice_list[index2]==[]):
        index2=random.randint(0,3)
    mouse2=mice_list[index2].pop()
    str_competition=mouse1+mouse2
    isostr_competition=mouse2+mouse1
    if (str_competition not in competition_list) and (isostr_competition not in competition_list):
        competition_list.append(str_competition)
    else:
        mice_list[index1].append(mouse1)
        mice_list[index2].append(mouse2)
        continue
    counter+=1
print(competition_list)
```

```python
#与上面的不同之处在于添加了一些内容用于避免同一只小鼠连续比三场
#created by Gangning Liu
import random

mice_list=[["A","A","A"],["B","B","B"],["C","C","C"],["D","D","D"]]
counter=0
past_index=[]
competition_list=[]
while counter!=6:
    index1=random.randint(0,3)
    while (mice_list[index1]==[]) or (index1 in past_index):
        index1=random.randint(0,3)
    mouse1=mice_list[index1].pop()
    index2=random.randint(0,3)
    while (index2==index1) or (mice_list[index2]==[]) or (index2 in past_index):
        index2=random.randint(0,3)
    mouse2=mice_list[index2].pop()
    str_competition=mouse1+mouse2
    isostr_competition=mouse2+mouse1
    if (str_competition not in competition_list) and (isostr_competition not in competition_list):
        competition_list.append(str_competition)
    else:
        continue
    counter+=1
    if counter%2==1:
        past_index=[index1,index2]
    else:
        past_index=[]
print(competition_list)
```
